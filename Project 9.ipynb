```python
# Project
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

class SalesDataAnalyzer:
    def __init__(self):
        self.df = None

    # Load data
    def load_data(self,file):
          try:
              self.df = pd.read_csv(file)
              print("\n Dataset uploaded \n")
          except Exception as e:
              print(e)
    # Explore data

    def explore_data(self):
            if self.df is not None:
                print("\n  data\n:")
                print(self.df.head())
    
                print("\ndata info:\n")
                print(self.df.info())
    
                print("\n Data describe:\n")
                print(self.df.describe())

                
    def clean_data(self):
        if self.df is not None:
            
            self.df.fillna(self.df.select_dtypes(include=np.number).mean(),inplace=True)
            
            print("\n clean successfully\n")
            print(self.df.head(8))
            
    
    def mathematical_operations(self):
        if self.df is not None:
            
            num = self.df.select_dtypes(include=np.number)

            print("\nAddition:")
            print(num + 5.5)

            print("\nSubtraction\n:")
            print(num - 10)

            print("\nmultipication:")
            print(num * 2)

            
    def search_data(self,column,value):
        if self.df is not None:
            
            result = self.df[self.df[column]==value]
            print(result)

            
    def sort_data(self,column):
        if self.df is not None:
            
            sort_dt = self.df.sort_values(by=column)
            print(sort_dt)

    def filter_data(self,column,value):
        if self.df is not None:
            fil = self.df[self.df[column] > value]
            print(fil)

    def aggregate_data(self):
        if self.df is not None:

            num = self.df.select_dtypes(include=np.number)

            print("\nSum:\n",num.sum())
            print("\nMean:\n",num.mean())
            print("\nCount:\n",num.count())

    def stat_data(self):
        if self.df is not None:
            num = self.df.select_dtypes(include=np.number)

            print("\nStandarded devi.:",num.std())
            print("\nVariance:",num.var())
            print("\nQuantile:",num.quantile([0.25,0.5,0.75]))

    def pivot_data(self,index_col,value_col):
        if self.df is not None:
            piv = pd.pivot_table(self.df,index=index_col,values=value_col,aggfunc="sum")
            print(piv)

    def bar_chart(self,x,y):
        if self.df is not None:
            top = self.df.sort_values(by=y,ascending=False).head(5)
            plt.bar(top[x],top[y],color="red",label="Bar chart")
            plt.legend()
            plt.xlabel(x)
            plt.ylabel(y)
            plt.show()
    
    def line_chart(self,x,y):
        if self.df is not None:
            plt.figure(figsize=(18,6))
            top = self.df.sort_values(by=y,ascending=False).head(10)
            plt.plot(top[x],top[y],color="blue",label="Line Plot")
            plt.legend()
            plt.xlabel(x)
            plt.ylabel(y)
            plt.show()

    def scatter_chart(self,x,y):
        if self.df is not None:
            top = self.df.sort_values(by=y,ascending=False)
            plt.scatter(top[x],top[y],color="green",edgecolor="black",label="Scatter Plot")
            plt.legend()
            plt.xlabel(x)
            plt.ylabel(y)
            plt.show()

    def pie_chart(self,x,y):
        if self.df is not None:
            top = self.df.groupby(x)[y].sum()
            plt.pie(top,labels=top.index,autopct="%1.1f%%")
            plt.legend()
            plt.xlabel(x)
            plt.ylabel(y)
            plt.savefig("piechart.png")
            plt.show()

    def sub_plot(self):
        plt.figure(figsize=(10,10))
#PIE CHART
        print("Pie chart")
        x = input("Enter the X-axis Column :")
        y = input("Enter the Y-axis Column :")
        top = self.df.groupby(x)[y].sum()
        
        plt.subplot(2,2,1)
        
        plt.pie(top,labels=top.index,autopct="%1.1f%%")
        plt.title("Pie chart")
        plt.legend()
        
        plt.xlabel(x)
        plt.ylabel(y)
        
#SCATTER PLOT
        print("Scatter chart")
        x1 = input("Enter the X-axis Column :")
        y1 = input("Enter the Y-axis Column :")
        top = self.df.sort_values(by=y1,ascending=False)
        
        plt.subplot(2,2,2)
        plt.scatter(top[x1],top[y1],color="pink",edgecolor="black",label="Scatter plot")
        plt.title("Scatter chart")
        plt.xlabel(x1)
        plt.ylabel(y1)
        plt.legend()

#BAR PLOT
        print("Bar chart")
        x2 = input("Enter the X-axis Column :")
        y2 = input("Enter the Y-axis Column :")
        top = self.df.sort_values(by=y2,ascending=False).head(5)

        plt.subplot(2,2,3)
        plt.bar(top[x2],top[y2],color="red",label="BAR")
        plt.legend()
        plt.title("BAR PLOT")
        plt.xlabel(x2)
        plt.ylabel(y2)
#HISTOGRAM CHART
        print("Hist chat")
        x3 = input("Enter the X-axis Column :")
        y3 = input("Enter the Y-axis Column :")
        top = self.df.sort_values(by=y3,ascending=False)

        plt.subplot(2,2,4)
        plt.hist(top[x3],bins=10,color="yellow",label="hist")
        plt.legend()
        plt.title("Hist Chart")
        plt.xlabel(x3)
        plt.ylabel(y3)
        plt.savefig("sub_plot.pdf")
        plt.show()

    def heatmap(self):

        sns.heatmap(self.df.corr(numeric_only=True),annot=True)
        plt.title("Heatmap")
        plt.show()


    def box_plot(self,columns):
        sns.boxplot(y=self.df[columns])
        plt.title("Box plot")
        plt.show()


        
def main():
        D = SalesDataAnalyzer()

        while True:


            print("\n========== Data Analysis & Visualization ==========")
    
            print("1. Load Dataset")
            print("2. Explore Data")
            print("3. Handle Missing Data")
            print("4. Mathematical Operations")
            print("5. Search Data")
            print("6. Sort Data")
            print("7. Filter Data")
            print("8. Aggregate Functions")
            print("9. Statistical Analysis")
            print("10. Create Pivot Table")
            print("11. Bar Plot")
            print("12. Line Plot")
            print("13. Scatter Plot")
            print("14. Pie Chart")
            print("15. Sub Plot")
            print("16. Heatmap")
            print("17. Box Plot")
            print("18. Exit")


            choice = int(input("Enter your choice: "))

            if choice == 1:
                # path = pd.read_csv("data.csv")
                D.load_data("data.csv")

                
            elif choice == 2:
                D.explore_data()

                
            elif choice == 3:
                D.clean_data()

                
            elif choice == 4:
                D.mathematical_operations()

                
            elif choice == 5:
                
                column = input("Enter data(col) for search:")
                value = float(input("Enter data(val) for search:"))
                D.search_data(column,value)

                
            elif choice == 6:
                
                column = input("Enter Column Name: ")
                D.sort_data(column)

            elif choice == 7:
                
                column = input("Enter Columns Name For Filter:")
                value = float(input("Enter Value For Filter:"))
                D.filter_data(column,value)

            elif choice == 8:

                D.aggregate_data()

            elif choice == 9:

                D.stat_data()

            elif choice == 10:

                index_col = input("Enter Index Col:")
                value_col = input("Enter Value Col:")               
                D.pivot_data(index_col,value_col)

            elif choice == 11:

                x = input("Enter the X-axis Column :")
                y = input("Enter the Y-axis Column :")
                print("\n ONLY FIRST 5 VALUE DATA \n")
                D.bar_chart(x,y)

            elif choice == 12:

                x = input("Enter the X-axis Column :")
                y = input("Enter the Y-axis Column :")
                print("\n ONLY FIRST 10 VALUE DATA \n")
                D.line_chart(x,y)

            elif choice == 13:

                x = input("Enter the X-axis Column :")
                y = input("Enter the Y-axis Column :")
                print("\n ONLY FIRST 10 VALUE DATA \n")
                D.scatter_chart(x,y)

            elif choice == 14:

                x = input("Enter the X-axis Column :")
                y = input("Enter the Y-axis Column :")
                print("\n ONLY FIRST 10 VALUE DATA \n")
                D.pie_chart(x,y)

            elif choice == 15:
                D.sub_plot()
                plt.savefig("sub_plot.pdf")

            elif choice == 16:
                D.heatmap()
                    
            elif choice == 17:
                col = input("Enter Column Name:")
                D.box_plot(col)
            else:
                print("Enter Valid Input")
                break

main()
```
